# Demo Linear Algebra: MiniNumPy vs NumPy Comparison

```python
import MiniNumPy as mnp
import numpy as np
import time
import random
```

## Matrix/Vector Operations

### Dot product

```python
def print_two_column_arrays(title, mn_arr, np_arr):
    print("=" * 60)
    print(title)
    print("MiniNumPy Array".ljust(30) + " | " + "NumPy Array")
    print("-" * 60)

    mn_lines = str(mn_arr).split("\n")
    np_lines = str(np_arr).split("\n")

    max_lines = max(len(mn_lines), len(np_lines))

    if len(mn_lines) < max_lines:
        mn_lines += [""] * (max_lines - len(mn_lines))
    if len(np_lines) < max_lines:
        np_lines += [""] * (max_lines - len(np_lines))

    for left, right in zip(mn_lines, np_lines):
        print(left.ljust(30) + " | " + right)

    print()

def print_two_column_scalar(title, mn_val, np_val):
    print("=" * 60)
    print(title)
    print(f"{'MiniNumPy':<20} | NumPy")
    print("-" * 60)
    print(f"{str(mn_val):<20} | {np_val}")
    print()
```

```python
v1 = [1, 2, 3]
v2 = [4, 5, 6]

M1 = [
    [1, 2, 3],
    [4, 5, 6]
]

M2 = [
    [7,  8],
    [9, 10],
    [11,12]
]

mnv1 = mnp.array(v1)
mnv2 = mnp.array(v2)

mnM1 = mnp.array(M1)
mnM2 = mnp.array(M2)

npv1 = np.array(v1)
npv2 = np.array(v2)

npM1 = np.array(M1)
npM2 = np.array(M2)
```

```python
print_two_column_scalar(
    "CASE 1: 1D dot 1D",
    mnp.dot(mnv1, mnv2),
    np.dot(npv1, npv2)
)

print_two_column_arrays(
    "CASE 2: 1D dot 2D",
    mnp.dot(mnv1, mnM2),
    np.dot(npv1, npM2)
)

print_two_column_arrays(
    "CASE 3: 2D dot 1D",
    mnp.dot(mnM1, mnv1),
    np.dot(npM1, npv1)
)

print_two_column_arrays(
    "CASE 4: 2D dot 2D",
    mnp.dot(mnM1, mnM2),
    np.dot(npM1, npM2)
)
```

### General matrix multiplication

```python
# 2D matrices
mn_A = mnp.arange(6).reshape((2,3))    # [[0,1,2],[3,4,5]]
mn_B = mnp.arange(6,12).reshape((3,2))

np_A = np.arange(6).reshape((2,3))
np_B = np.arange(6,12).reshape((3,2))

# 3D
mn_A3 = mnp.arange(24).reshape((2,3,4))
mn_B3 = mnp.arange(48).reshape((4,12))

np_A3 = np.arange(24).reshape((2,3,4))
np_B3 = np.arange(48).reshape((4,12))

# BROADCAST batch matmul
mn_A_b = mnp.arange(12).reshape((3,4))
mn_B_b = mnp.arange(40).reshape((5,4,2))

np_A_b = np.arange(12).reshape((3,4))
np_B_b = np.arange(40).reshape((5,4,2))
```

```python
print_two_column_arrays(
    "matmul CASE4: 2D @ 2D",
    mnp.matmul(mn_A, mn_B),
    np.linalg.matmul(np_A, np_B)
)

print_two_column_arrays(
    "matmul CASE5: 3D @ 3D",
    mnp.matmul(mn_A3, mn_B3),
    np.linalg.matmul(np_A3, np_B3)
)

print_two_column_arrays(
    "matmul CASE6: broadcast batch",
    mnp.matmul(mn_A_b, mn_B_b),
    np.linalg.matmul(np_A_b, np_B_b)
)
```

### Norm

```python
# N-D datasets (ndim != 2)
ND_tests = [
    ("1D v1",  mnp.arange(5),              np.arange(5)),
    ("1D v2",  mnp.arange(-3, 2),          np.arange(-3, 2)),
    ("3D C3",  mnp.arange(24).reshape((2,3,4)),  np.arange(24).reshape((2,3,4))),
    ("4D D4",  mnp.arange(48).reshape((2,2,3,4)),np.arange(48).reshape((2,2,3,4))),
]
```

```python
for title, mn_x, np_x in ND_tests:
    print_two_column_scalar(
        f"{title} — norm(None/2)",
        mnp.norm(mn_x),
        np.linalg.norm(np_x)
    )

    print (mnp.norm(mn_x, 1))
    print (mnp.norm(mn_x, 3))
    
```

```python
# 2D datasets (ndim == 2)
D2_tests = [
    ("2D A",  mnp.arange(6).reshape((2,3)),     np.arange(6).reshape((2,3))),
    ("2D B",  mnp.arange(20).reshape((4,5)), np.arange(20).reshape((4,5))),
]
```

```python
for title, mn_M, np_M in D2_tests:
    print_two_column_scalar(
        f"{title} — norm(None/Fro)",
        mnp.norm(mn_M),
        np.linalg.norm(np_M)
    )

    print_two_column_scalar(
        f"{title} — norm(1) (max column sum)",
        mnp.norm(mn_M, 1),
        np.linalg.norm(np_M, 1)
    )
    
    print_two_column_scalar(
        f"{title} — norm(2) (spectral)",
        mnp.norm(mn_M, 2),
        np.linalg.norm(np_M, 2)
    )

    print_two_column_scalar(
        f"{title} — norm(3) (max row sum)",
        mnp.norm(mn_M, 3),
        np.linalg.norm(np_M, np.inf)
    )
```

### Determinant

```python
det_tests = [
    ("1x1", mnp.arange(1).reshape((1,1)), np.arange(1).reshape((1,1))),
    
    ("2x2 simple", mnp.arange(4).reshape((2,2)), np.arange(4).reshape((2,2))),
    
    ("2x2 mixed", mnp.array([[3, -1], [2, 5]]), np.array([[3, -1], [2, 5]])),
    
    ("3x3", mnp.arange(9).reshape((3,3)), np.arange(9).reshape((3,3))),
    
    ("3x3 mixed", 
        mnp.array([[2, 4, 1], [3, -1, 5], [7, 0, 2]]),
        np.array([[2, 4, 1], [3, -1, 5], [7, 0, 2]])
    ),

    ("4x4", mnp.arange(16).reshape((4,4)), np.arange(16).reshape((4,4))),
]
```

```python
for title, mn_M, np_M in det_tests:
    print_two_column_scalar(
        f"{title} — det",
        mnp.det(mn_M),
        np.linalg.det(np_M)
    )
```

### Inverse

```python
inv_tests = [
    ("2x2 simple", 
        mnp.array([[1, 2], [3, 4]]),
        np.array([[1, 2], [3, 4]])),

    ("2x2 mixed",
        mnp.array([[4, 7], [2, 6]]),
        np.array([[4, 7], [2, 6]])),

    ("3x3 basic",
        mnp.arange(9).reshape((3,3)) + mnp.ones((3,3)),   # ensure invertible
        (np.arange(9).reshape((3,3)) + 1)
    ),

    ("3x3 random",
        mnp.array([[2, 4, 1], [3, 5, 7], [1, 0, 6]]),
        np.array([[2, 4, 1], [3, 5, 7], [1, 0, 6]])
    ),

    # Singular (det = 0) — expect error
    ("singular", 
        mnp.array([[1, 2], [2, 4]]),
        np.array([[1, 2], [2, 4]])
    ),
]
```

```python
for title, mn_M, np_M in inv_tests:
    print("="*60)
    print(title)

    try:
        mn_inv = mnp.inv(mn_M)
    except Exception as e:
        mn_inv = f"Error: {type(e).__name__}"

    try:
        np_inv = np.linalg.inv(np_M)
    except Exception as e:
        np_inv = f"Error: {type(e).__name__}"

    if isinstance(mn_inv, str):
        print("MiniNumPy:", mn_inv)
        print("NumPy    :", np_inv)
        print()
    else:
        print_two_column_arrays(title + " — inverse", mn_inv, np_inv)
```

### eig

## Factorizations and Solver

### LU decomposition

```python
def random_float_matrix(n, low=-10.0, high=10.0):
    data = [[random.uniform(low, high) for _ in range(n)] for _ in range(n)]
    return mnp.array(data)

T = random_float_matrix(4)
print("Random Matrix T:")
print(T)
P, L, U = T.LU_Decomposition()

print("Permutaion matrix:")
print(P)
print("Lower triagular matrix")
print(L)
print("Upper triagular matrix")
print(U)
print("="* 30 + "Validation Result" + "="*30)
print(L@U - P@T)
```

### QR decomposition

```python
T1 = random_float_matrix(4)
print("Random Matrix T1:")
print(T1)

Q, R = mnp.qr_decomposition(T)

print("Q matrix:")
print(Q)
print("R matrix:")
print(R)
print("="* 30 + "Validation Result" + "="*30)
print(Q@R - T)
```

### Unique solution solver

```python
# solve AX = B
Sa = [
    [ 2,  1, -1],
    [-3, -1,  2],
    [-2,  1,  2]
]

Sb = [
    [8],
    [-11],
    [-3]
]

mnA = mnp.array(Sa)
mnB = mnp.array(Sb) 
npA = np.array(Sa)
npB = np.array(Sb)

mn_x = mnp.solve(mnA, mnB)
np_x = np.linalg.solve(npA, npB)

print_two_column_arrays(
    "Solving AX = B",
    mn_x,
    np_x
)
```
